```python
df = pd.read_csv('Position_Salaries.csv')
```

```python
X = df.iloc[:,1:2].values
y = df.iloc[:,[2]].values
```

```python
#becuse we have small data so we dont need to do split data train and test its not important
```

```python
from sklearn.preprocessing import StandardScaler
sc_X = StandardScaler()
```

```python
sc_y = StandardScaler()
```

```python
X = sc_X.fit_transform(X)
```

```python
y = np.ravel(sc_y.fit_transform(y.reshape(-1,1)))
# reval will do to return my y to orgenal shap
# will do in code is to change y chape to 2d (10,1) and return to 1d (10,)
```

```python
X
```

```python
from sklearn.svm import SVR
```

```python
regressor = SVR(kernel='linear')
```

```python
regressor
```

```python
regressor.fit(X,y)
```

```python
y_pred = sc_y.inverse_transform(regressor.predict(sc_X.transform([[3.5]])).reshape(-1,1))
```
